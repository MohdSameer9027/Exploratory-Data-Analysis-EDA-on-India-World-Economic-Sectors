### Visualization

## Bar Chart

```python
# Country_wise GDP

plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="Country ",
    y="GDP_Value_USD_Bn",
    estimator=np.mean 
)
plt.xticks(rotation=45)
plt.title("Average GDP by Country")
plt.show()
```

## Pie Chart

```python
# sector Distribution

df["Sector "].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)
plt.title("Sector Distribution")
plt.show()
```

## Histogram

```python
# GDP Distribution

plt.figure(figsize=(8,5))
plt.hist(df["GDP_Value_USD_Bn"],bins=20)
plt.title("GDP Distribution")
plt.show()
```

```python
# Histogram with Kde

sns.histplot(
    df["GDP_Value_USD_Bn"],
    kde=True
)
```

## Box plot

```python
#Purpose Detect Outliers

sns.boxplot(
    y=df["GDP_Value_USD_Bn"]
)
```

## Scatter Plot

## Heatmap
